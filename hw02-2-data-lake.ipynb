# Part 2.2: Storing OLTP Events in a Data Lake

```python
from pathlib import Path

import boto3

NETID = "yg452"
MY_BUCKET = f"dsan6000-{NETID}"
DEST_PREFIX = "wikipedia-hourly"
DATA_DIR = Path("data")

s3 = boto3.client("s3")
```

```python
s3.create_bucket(Bucket=MY_BUCKET)
print("created", f"s3://{MY_BUCKET}")
```

```python
local_files = sorted(DATA_DIR.glob("*.parquet"))

for f in local_files:
    s3.upload_file(str(f), MY_BUCKET, f"{DEST_PREFIX}/{f.name}")

print(len(local_files), "files uploaded")
```

```python
resp = s3.list_objects_v2(Bucket=MY_BUCKET, Prefix=f"{DEST_PREFIX}/")
for obj in resp["Contents"]:
    print(f"{obj['Size']:>10,}  s3://{MY_BUCKET}/{obj['Key']}")
```
